```python
import overpy
import pandas as pd
import json
from collections import Counter
```

```python
api = overpy.Overpass()
```

```python
#サンプルエリア：浅草周辺
query = """
[out:json][timeout:60];
(
  node["tourism"~"museum|attraction|gallery|theme_park|viewpoint|zoo|aquarium"](35.710,139.790,35.720,139.800);
  way["tourism"~"museum|attraction|gallery|theme_park|viewpoint|zoo|aquarium"](35.710,139.790,35.720,139.800);
  relation["tourism"~"museum|attraction|gallery|theme_park|viewpoint|zoo|aquarium"](35.710,139.790,35.720,139.800);
);
out body;
>;
out skel qt;
"""
```

```python
print("クエリ実行中...")
result = api.query(query)

# すべてのタグを収集
all_tags = []
for element in result.nodes + result.ways + result.relations:
    all_tags.extend(element.tags.keys())

# タグの出現頻度
tag_counts = Counter(all_tags)

print(f"\n取得した観光地数: {len(result.nodes) + len(result.ways) + len(result.relations)}")
print(f"\nユニークなタグ数: {len(tag_counts)}")
print("\n=== 頻出タグ TOP 20 ===")
for tag, count in tag_counts.most_common(20):
    print(f"{tag}: {count}")

# バリアフリー関連タグを抽出
bf_tags = {k: v for k, v in tag_counts.items() 
           if any(keyword in k.lower() for keyword in 
                  ['wheelchair', 'elevator', 'ramp', 'toilet', 'door', 
                   'access', 'tactile', 'description'])}

print("\n=== バリアフリー関連タグ ===")
for tag, count in sorted(bf_tags.items(), key=lambda x: x[1], reverse=True):
    print(f"{tag}: {count}")

# サンプルデータの詳細確認
print("\n=== サンプル観光地の全タグ ===")
for i, node in enumerate(result.nodes[:3]):  # 最初の3件
    print(f"\n--- 観光地 {i+1}: {node.tags.get('name', 'Unknown')} ---")
    for key, value in node.tags.items():
        print(f"  {key}: {value}")
```

```python
print(f"観光地（ノード）: {len(result.nodes)}")
print(f"観光地（ウェイ）: {len(result.ways)}")
print(f"観光地（リレーション）: {len(result.relations)}")
print(f"合計: {len(result.nodes) + len(result.ways) + len(result.relations)}")

# 観光地本体のみカウント
tourism_count = sum(1 for n in result.nodes if 'tourism' in n.tags)
tourism_count += sum(1 for w in result.ways if 'tourism' in w.tags)
tourism_count += sum(1 for r in result.relations if 'tourism' in r.tags)
print(f"\ntourismタグを持つ要素: {tourism_count}")
```
